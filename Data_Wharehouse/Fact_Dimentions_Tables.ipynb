#data wharehousing 

```sql
CREATE DATABASE orderDWH
```

###staging layer

```sql
--staging layer
--initial load

Create or replace table orderDWH.stg_sales
as
select * from sales_new.Orders
where OrderDate > '2024-02-20';
```

##transformation


```sql
create view orderDWH.trans_sales
AS
select * from orderDWH.stg_sales where Quantity is not null;
```

```sql
select * from  orderDWH.trans_sales
```

##core layer

#### dimcust

```sql
create or replace table orderdwh.dimcustomer(
    CustomerId int,
    Dimcustkey int,
    CustomerName string,
    CustomerEmail string
)
```

```sql
    create or replace view orderdwh.view_dimcust
    AS 
    select T.*,row_number() over(order by CustomerID) as Dimcustkey from 
    (
        select 
            distinct(CustomerID) AS CustomerId,
            CustomerName,
            CustomerEmail
        from 
            orderdwh.trans_sales
    )AS T
```

```sql
insert into orderdwh.dimcustomer
select CustomerId, Dimcustkey, CustomerName, CustomerEmail from orderdwh.view_dimcust
```

#### product dim

```sql
 create or replace table orderdwh.dimproduct(
    pid int,
    pname string,
    pcatagory string,
    DimPkey int
)
```

```sql
    create or replace view orderdwh.view_dimproduct
    AS 
    select T.*,row_number() over(order by ProductID) as DimPkey from 
    (
        select 
            distinct(ProductID) AS ProductID,
            ProductName,
            ProductCategory
        from 
            orderdwh.trans_sales
    )AS T
```

```sql
insert into orderdwh.dimproduct
select * from orderdwh.view_dimproduct
```

```sql
select * from orderdwh.dimproduct
```

####region dim


```sql
create or replace table orderdwh.dimregion(
    RegionId int,
    RegionName string,
    Country string,
    dimregionkey int  
)
```

```sql
     create or replace view orderdwh.view_dimpregion
    AS 
    select T.*,row_number() over(order by RegionId) as dimregionkey from 
    (
        select 
            distinct(RegionId) AS RegionId,
            RegionName,
            Country
        from 
            orderdwh.trans_sales
    )AS T
```

```sql
insert into orderdwh.dimregion
select * from orderdwh.view_dimpregion
```

```sql
select * from orderdwh.dimregion
```

####dimdate


```sql
 create or replace table orderdwh.dimdate(
    OrderDate Date,
    dimdatekey int
)   
```

```sql
    create or replace view orderdwh.view_dimdate
    AS 
    select T.*,row_number() over(order by OrderDate) as dimdatekey from 
    (
        select 
            distinct(OrderDate) AS OrderDate
        from 
            orderdwh.trans_sales
    )AS T
```

```sql
insert into orderdwh.dimdate
select * from orderdwh.view_dimdate
```

###fact table


```sql
Create or replace Table orderdwh.Facttbl(
    OrderId int,
    Quentity decimal,
    UnitPrice decimal,
    TotalAmount decimal,
    dimcustkey int,
    dimproductkey int,
    dimregionkey int,
    dimdatekey int

)
```

```sql
select
    F.orderId,
    F.Quantity,
    F.UnitPrice,
    F.TotalAmount,
    DC.dimcustkey,
    DP.DimPkey,
    DR.dimregionkey,
    DD.dimdatekey
from
    orderdwh.trans_sales F
left join
    orderdwh.dimcustomer DC
    ON F.CustomerID = DC.CustomerId
left join
    orderdwh.dimproduct DP
    ON F.ProductID = DP.pid
left join
    orderdwh.dimregion DR
    ON DR.Country = f.Country
left join
    orderdwh.dimdate DD
    ON F.OrderDate = DD.OrderDate
```
